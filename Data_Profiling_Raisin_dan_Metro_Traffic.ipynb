identifikasi outlier

```python
# outlier Raisin menggunakan IQR
Q1_r = numeric_raisin.quantile(0.25)
Q3_r = numeric_raisin.quantile(0.75)
IQR_r = Q3_r - Q1_r
outlier_raisin = ((numeric_raisin < (Q1_r - 1.5 * IQR_r)) | (numeric_raisin > (Q3_r + 1.5 * IQR_r)))
print("Jumlah Outlier Raisin per kolom:\n", outlier_raisin.sum())
```

```python
# outlier Traffic menggunakan IQR
numeric_traffic = df_traffic.select_dtypes(include=np.number)
Q1_t = numeric_traffic.quantile(0.25)
Q3_t = numeric_traffic.quantile(0.75)
IQR_t = Q3_t - Q1_t
outlier_traffic = ((numeric_traffic < (Q1_t - 1.5 * IQR_t)) | (numeric_traffic > (Q3_t + 1.5 * IQR_t)))
print("Jumlah Outlier Metro Traffic per kolom:\n", outlier_traffic.sum())
```

timeliness khusus metro traffic

```python
# konversi ke datetime
df_traffic['date_time'] = pd.to_datetime(df_traffic['date_time'], errors='coerce')

print("Tipe data date_time sekarang:", df_traffic['date_time'].dtype)
print("Tanggal awal:", df_traffic['date_time'].min())
print("Tanggal akhir:", df_traffic['date_time'].max())

# Cek Time Gap (Kesinambungan Waktu)
df_traffic_time = df_traffic.sort_values('date_time').copy()
df_traffic_time['time_gap'] = df_traffic_time['date_time'].diff()

gaps = df_traffic_time[df_traffic_time['time_gap'] > pd.Timedelta(hours=1)]
print("\nJumlah gap > 1 jam:", len(gaps))
print("Gap terbesar:", df_traffic_time['time_gap'].max())

display(gaps[['date_time', 'time_gap']].head())
```
